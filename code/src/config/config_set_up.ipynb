```python
base = "dbfs:/Volumes/dev/infra/pipeline_artifacts"

extra_data = {
    "checkpoint_path": f"{base}/checkpoint",
    "logs_path": f"{base}/logs",
    "landing_area": f"{base}/landing_area",
    "watermark_path": f"{base}/watermark",
}

# Drop (recursively) if exists, then recreate
for name, path in extra_data.items():
    try:
        # Try deleting regardless of existence; ignore errors for missing paths
        dbutils.fs.rm(path, True)
        print(f"Deleted : {path}")
    except Exception:
        print(f"Skipped delete (not found): {path}")

for name, path in extra_data.items():
    dbutils.fs.mkdirs(path)
    print(f"Created: {path}")
```
